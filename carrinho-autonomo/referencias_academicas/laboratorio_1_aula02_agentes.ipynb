### A primeira célula: Quem está rodando meu código?

A célula abaixo é um teste básico. Ela apenas **pergunta à máquina remota quais são suas configurações** (versão do Python e sistema operacional). 

Usamos duas bibliotecas nativas do Python para isso:
- `sys`: Traz informações sobre o interpretador Python.
- `platform`: Traz dados do sistema operacional.

````python
# ---------------------------------------------------------------------
# Objetivo desta célula: descobrir em que máquina e em que versão do
# Python o nosso código está rodando. É a primeira coisa que se checa
# quando algo dá errado ("funciona na minha máquina" costuma ser um
# problema de versão diferente).
# ---------------------------------------------------------------------

import sys        # módulo com informações sobre o interpretador Python
import platform   # módulo com informações sobre o sistema operacional

# sys.version é um texto longo, algo como "3.11.8 (main, Feb 2024) [GCC 11]".
# O método .split() quebra esse texto em pedaços separados por espaço e
# devolve uma lista; [0] pega o primeiro pedaço, que é só o número da versão.
versao_python = sys.version.split()[0]

# platform.system() devolve o nome do sistema ("Linux", "Windows", "Darwin").
# platform.release() devolve a versão do sistema.
sistema = platform.system()
versao_sistema = platform.release()

# A função print() exibe valores na tela. Quando passamos vários valores
# separados por vírgula, o Python os imprime lado a lado, separados por espaço.
print("Versão do Python:", versao_python)
print("Sistema:", sistema, versao_sistema)
````

### A volatilidade do Runtime (e por que isso afeta sua nota)

### Tipos Básicos (Os blocos elementares)

No Python, o tipo da variável é definido automaticamente pelo valor que você atribui a ela. 
- `str`: Texto (sempre entre aspas).
- `bool`: Verdadeiro ou Falso (`True` / `False`).
- `float`: Número decimal (usando ponto, não vírgula).
- `int`: Número inteiro.

````python
# ---------------------------------------------------------------------
# Os quatro tipos básicos que usaremos. A função type() revela o tipo
# de qualquer valor — útil quando um cálculo dá resultado estranho e
# você suspeita que uma variável não é o que você pensava.
# ---------------------------------------------------------------------

local = "A"          # str   -> texto entre aspas
esta_sujo = True     # bool  -> só aceita True ou False
custo = 1.5          # float -> número real (repare no PONTO, não vírgula)
passos = 10          # int   -> número inteiro

# Note que o Python entendeu o tipo sozinho, sem declararmos nada.
print("local     =", local, "| tipo:", type(local))
print("esta_sujo =", esta_sujo, "| tipo:", type(esta_sujo))
print("custo     =", custo, "| tipo:", type(custo))
print("passos    =", passos, "| tipo:", type(passos))
````

### Tupla: A estrutura da Percepção

Uma **tupla** é criada usando parênteses `()`. A sua principal característica é ser **imutável**: uma vez criada, não pode ser alterada.

Por isso ela representa tão bem uma **percepção** em IA. Se o sensor do agente leu que a sala está suja naquele milissegundo, esse fato é histórico e imutável. 

**Dica de sintaxe: Desempacotamento.**
Podemos extrair os valores da tupla direto para variáveis:
```python
local, estado = ("A", "Sujo")
```

````python
# ---------------------------------------------------------------------
# TUPLA — usada para representar uma PERCEPÇÃO.
# ---------------------------------------------------------------------

# Criamos a tupla: dois valores entre parênteses, separados por vírgula.
percepcao = ("A", "Sujo")

print("A percepção completa:", percepcao)
print("Quantos elementos ela tem?", len(percepcao))   # len() conta os elementos

# Acesso por POSIÇÃO. Atenção: em Python a contagem começa em ZERO,
# então o primeiro elemento é [0] e o segundo é [1].
print("Elemento na posição 0:", percepcao[0])
print("Elemento na posição 1:", percepcao[1])

# DESEMPACOTAMENTO: quebra a tupla em duas variáveis de uma só vez.
# À esquerda do "=" colocamos tantos nomes quantos forem os elementos.
local_atual, estado_atual = percepcao

print("Depois do desempacotamento -> local:", local_atual, "| estado:", estado_atual)

# A tupla é IMUTÁVEL: a linha abaixo daria erro se fosse executada.
# Descomente para ver o TypeError acontecer — errar de propósito ensina.
# percepcao[0] = "B"
````

### Lista: A estrutura das Ações e Históricos

As **listas** usam colchetes `[]`. Diferente da tupla, elas são **mutáveis**. Você pode adicionar, remover e alterar itens dinamicamente.

Usaremos listas para registrar as **ações disponíveis** (que podem mudar dependendo do cenário) e os **históricos** de navegação do agente.

**Principais comandos:**
- `lista.append(x)`: Adiciona `x` ao final da lista.
- `lista[0]`: Acessa o primeiro elemento.
- `lista[-1]`: Acessa o último elemento.
- `len(lista)`: Retorna o tamanho da lista.
- `x in lista`: Verifica se `x` existe dentro da lista.

````python
# ---------------------------------------------------------------------
# LISTA — usada para o conjunto de AÇÕES e para históricos.
# ---------------------------------------------------------------------

# Criamos a lista com três ações. Colchetes, e não parênteses.
acoes = ["Aspirar", "Direita", "Esquerda"]
print("Ações iniciais:", acoes)

# append() ACRESCENTA um elemento ao final. Repare que a lista MUDA:
# é a diferença fundamental em relação à tupla.
acoes.append("NadaFazer")
print("Depois do append:", acoes)

# Acesso por posição, igual à tupla: começa em zero.
print("Primeira ação :", acoes[0])

# Índice NEGATIVO conta de trás para frente: -1 é o último elemento.
# Isso evita escrever acoes[len(acoes) - 1], que é mais longo e mais frágil.
print("Última ação   :", acoes[-1])

print("Total de ações:", len(acoes))

# O operador "in" verifica se um valor está na lista. Devolve True ou False.
# É assim que o agente pergunta: "esta ação é permitida?"
print("'Aspirar' é uma ação válida?", "Aspirar" in acoes)
print("'Voar' é uma ação válida?   ", "Voar" in acoes)
````

### Dicionário: O Estado do Mundo

Os **dicionários** são criados com chaves `{}` e funcionam com o sistema de **Chave → Valor**. 

Eles são a melhor forma de representar o **Estado do Mundo**. Podemos mapear diretamente os locais do ambiente para as suas condições físicas:
```python
mundo = {"A": "Sujo", "B": "Limpo"}
```

**Acesso inteligente:** Em vez de decorar números (índices), acessamos os valores pelos nomes das chaves (`mundo["A"]`). Isso torna o código extremamente fácil de ler.

````python
# ---------------------------------------------------------------------
# DICIONÁRIO — usado para representar o ESTADO DO MUNDO.
#
# Estrutura: {chave: valor, chave: valor}
# Aqui, cada chave é um local do ambiente e cada valor é a condição
# daquele local. Este dicionário é, literalmente, o estado do mundo
# do aspirador num dado instante.
# ---------------------------------------------------------------------

mundo = {"A": "Sujo", "B": "Limpo"}
print("Estado do mundo:", mundo)

# LEITURA por chave: usamos o NOME entre colchetes, não um número.
print("O local A está:", mundo["A"])
print("O local B está:", mundo["B"])

# ESCRITA: atribuir a uma chave existente ALTERA o valor.
# É assim que a ação "Aspirar" modifica o ambiente.
mundo["A"] = "Limpo"
print("Depois de aspirar A:", mundo)

# .keys() devolve as chaves; .values() devolve os valores.
print("Locais existentes  :", list(mundo.keys()))
print("Condições dos locais:", list(mundo.values()))

# .get() consulta com segurança: devolve None se a chave não existir,
# em vez de interromper o programa com KeyError.
print("E o local C?", mundo.get("C"))
````

### Função: O Programa do Agente

Uma **função** agrupa um bloco de código que recebe entradas (parâmetros) e devolve uma saída. Em IA, o cérebro do nosso agente será uma função: ela recebe uma percepção, processa a lógica e devolve uma ação.

**Anatomia da Função:**
- Começa com a palavra `def`.
- É obrigatório **indentar** (recuar) as linhas dentro da função.
- O `return` finaliza a função entregando o resultado.

````python
# ---------------------------------------------------------------------
# FUNÇÃO — a estrutura que representa o PROGRAMA DO AGENTE.
#
# Esta primeira função é simples de propósito: ela apenas informa
# quanto custa executar cada ação. Usaremos isso mais adiante, quando
# a medida de desempenho passar a penalizar movimento.
# ---------------------------------------------------------------------

def custo_da_acao(acao):
    """Devolve o custo de executar uma ação: 0 para NadaFazer, 1 para as demais.

    Parâmetro:
        acao (str): o nome da ação, por exemplo "Aspirar".

    Devolve:
        int: o custo da ação.
    """
    # A comparação usa DOIS sinais de igual (==). Um sinal só (=) é
    # atribuição; dois são comparação. Trocar os dois é erro clássico.
    if acao == "NadaFazer":
        return 0        # return encerra a função AQUI; nada abaixo executa

    # Se a execução chegou nesta linha, é porque o if acima foi falso.
    return 1


# Chamando a função: escrevemos o nome e passamos o argumento entre parênteses.
print("Custo de 'Aspirar'  :", custo_da_acao("Aspirar"))
print("Custo de 'NadaFazer':", custo_da_acao("NadaFazer"))
````

### Laços de Repetição (`For`)

O comando `for` percorre os itens de uma lista ou dicionário. É ele que permitirá simularmos o agente andando pelo mundo repetidas vezes sem termos que copiar e colar código.

> **Dica visual (F-strings):** Usaremos muito o `f"texto {variavel}"` para facilitar a impressão de mensagens combinadas com dados.

````python
# ---------------------------------------------------------------------
# LAÇOS — percorrendo listas e dicionários.
# ---------------------------------------------------------------------

# Laço sobre uma LISTA: a variável "acao" assume um valor por vez.
print("Custo de cada ação:")
for acao in acoes:
    # f-string: o texto começa com f e as variáveis vão entre chaves.
    # :12s alinha o texto em 12 caracteres, deixando a saída em colunas.
    print(f"  {acao:12s} -> custo {custo_da_acao(acao)}")

print()   # print() sem argumentos apenas pula uma linha

# Laço sobre um DICIONÁRIO com .items(): recebemos chave e valor de uma vez.
print("Situação de cada local:")
for lugar, estado in mundo.items():
    print(f"  O local {lugar} está {estado}.")
````

# Etapa 3 — Pandas: Tabelas para Machine Learning

A partir da Unidade II, **tudo será uma tabela**. Dados de treino, testes e métricas vão depender da biblioteca Pandas. Introduzir essa ferramenta agora vai salvar muito tempo lá na frente.

Conceitualmente, no Pandas trabalharemos com o **estado fatorado**. Cada linha da tabela é um estado/observação, e cada coluna é uma variável que descreve esse estado.

## Estruturas Fundamentais
- **Series:** Uma única coluna de dados.
- **DataFrame:** A tabela inteira (várias Series organizadas lado a lado).

Vamos criar nossa primeira tabela convertendo dicionários em um `DataFrame`.

````python
# ---------------------------------------------------------------------
# Criando um DataFrame a partir de um dicionário de listas.
#
# Estrutura: pd.DataFrame({ "coluna1": [...], "coluna2": [...] })
#
# Esta tabela é o HISTÓRICO de um agente aspirador ao longo de 6 passos.
# Cada linha é um passo da simulação; cada coluna é um aspecto dele.
# ---------------------------------------------------------------------

import pandas as pd

registros = pd.DataFrame({
    "passo":  [1, 2, 3, 4, 5, 6],                                    # numeração do passo
    "local":  ["A", "A", "B", "B", "A", "B"],                        # onde o agente estava
    "estado": ["Sujo", "Limpo", "Sujo", "Limpo", "Limpo", "Limpo"],  # o que ele percebeu
    "acao":   ["Aspirar", "Direita", "Aspirar", "Esquerda",          # o que ele fez
               "Direita", "NadaFazer"],
    "custo":  [1, 1, 1, 1, 1, 0],                                    # quanto custou
})

# .head() mostra as primeiras linhas (5 por padrão). Em tabelas grandes,
# é o primeiro comando que se roda: serve para conferir se os dados
# entraram com a cara esperada, sem despejar milhares de linhas na tela.
registros.head()
````

### Inspecionando a Tabela (O Raio-X dos Dados)

Ao pegar qualquer Dataset na vida real, faça estas 3 perguntas:

1. **Tamanho:** `.shape` (linhas, colunas)
2. **Tipos:** `.dtypes` (o que é número, o que é texto)
3. **Estatísticas Básicas:** `.describe()` (médias, contagens, desvios). Identificar anomalias por aqui é essencial.

````python
# ---------------------------------------------------------------------
# As três perguntas iniciais sobre qualquer tabela.
# ---------------------------------------------------------------------

# 1) TAMANHO: devolve uma tupla (número de linhas, número de colunas).
#    Sem parênteses, porque shape é um atributo e não um método.
print("Formato (linhas, colunas):", registros.shape)
print()

# 2) COLUNAS e TIPOS. "object" é como o pandas chama colunas de texto;
#    "int64" é número inteiro.
print("Tipos de cada coluna:")
print(registros.dtypes)
print()

# 3) DISTRIBUIÇÃO das colunas numéricas.
#    Só "passo" e "custo" aparecem, porque as demais são texto.
print("Estatística descritiva:")
print(registros.describe())
````

### Selecionando Colunas

Funciona exatamente como os dicionários:
- `tabela["nome"]`: Retorna 1 coluna (tipo Series).
- `tabela[["nome1", "nome2"]]`: Retorna várias colunas (tipo DataFrame). Atenção aos **colchetes duplos** aqui!

````python
# ---------------------------------------------------------------------
# Selecionando colunas de um DataFrame.
# ---------------------------------------------------------------------

# UMA coluna, com um nome entre colchetes -> devolve uma Series.
coluna_acao = registros["acao"]

print("Tipo do resultado:", type(coluna_acao).__name__)
print(coluna_acao)
print()

# VÁRIAS colunas, com uma LISTA de nomes -> devolve um DataFrame.
# Colchetes duplos: os externos selecionam, os internos formam a lista.
subtabela = registros[["passo", "local", "acao"]]

print("Tipo do resultado:", type(subtabela).__name__)
print(subtabela)
````

### Filtros Booleanos (Cortando a tabela)

A operação de filtro no pandas acontece em 2 passos lógicos (mas escrevemos tudo em 1 linha só).

1. **Criação da máscara:** Ao fazer `registros["custo"] > 0`, o Pandas cria uma lista de Verdadeiros e Falsos.
2. **Aplicação da máscara:** Envelopamos essa condição dentro da própria tabela: `registros[registros["custo"] > 0]`.

> **Regra de Combinação:** No Pandas, use `&` para "E" e `|` para "OU", sempre colocando as condições menores dentro de parênteses.

````python
# ---------------------------------------------------------------------
# FILTRO BOOLEANO, mostrado em duas etapas para o mecanismo ficar claro.
# ---------------------------------------------------------------------

# ETAPA 1: a condição, sozinha, produz uma Series de True/False.
mascara = registros["custo"] > 0

print("A máscara (uma resposta por linha):")
print(mascara)
print()

# ETAPA 2: usamos a máscara para selecionar as linhas marcadas com True.
com_custo = registros[mascara]

print("As linhas selecionadas pela máscara:")
print(com_custo)
print()

# Na prática, as duas etapas costumam ser escritas juntas.
# É o mesmo resultado, em uma linha só:
print("Mesma coisa, escrito de forma compacta:")
print(registros[registros["custo"] > 0])
print()

# COMBINANDO condições: & significa "e". Cada condição vai entre parênteses,
# porque & tem precedência maior que > e faria a conta na ordem errada.
print("Passos no local A em que houve custo:")
print(registros[(registros["local"] == "A") & (registros["custo"] > 0)])
````

# Etapa 4 — O Mundo do Aspirador (Simulador)

Chegou a hora de aplicar toda essa infraestrutura na construção da nossa simulação de IA.

## Definição Formal do Ambiente

| Propriedade | Descrição |
|---|---|
| **Locais** | Dois: A (esquerda) e B (direita) |
| **Condição de Estado** | `"Limpo"` ou `"Sujo"` |
| **Percepção** | Ex: `("A", "Sujo")` |
| **Ações** | `"Aspirar"`, `"Direita"`, `"Esquerda"`, `"NadaFazer"` |

## Medida de Desempenho
A forma de avaliar o robô não é pelo esforço, mas pelo resultado.
> **Regra:** A cada passo de tempo, ganhe **+1 ponto para cada local limpo** no mapa.

Isso evita o "agente trapaceiro" que limpa, suja, e limpa novamente só para ganhar pontos de aspiração.

---

### Agente 1: Reativo Simples
Age sem memória, apenas com base na percepção imediata: *"Se está sujo, aspire; se não, vá para o outro lado."*

````python
# ---------------------------------------------------------------------
# AGENTE 1 — reativo simples.
#
# Decide olhando APENAS a percepção atual. Sem memória, sem planos.
# Esta função é o "programa do agente" da aula teórica: recebe uma
# percepção, devolve uma ação.
# ---------------------------------------------------------------------

def agente_reativo(percepcao, memoria=None):
    """Agente reativo simples: decide apenas com a percepção atual.

    Parâmetros:
        percepcao (tuple): o par (local, estado), ex. ("A", "Sujo").
        memoria (dict): não utilizado por este agente. Existe para que
            todos os agentes tenham a mesma assinatura.

    Devolve:
        str: a ação escolhida.
    """
    # Desempacota a percepção nas duas informações que ela carrega.
    local, estado = percepcao

    # REGRA 1: sujeira tem prioridade sobre qualquer outra coisa.
    if estado == "Sujo":
        return "Aspirar"

    # Se chegamos aqui, o local está limpo: resta se mover para o outro.
    # A sintaxe abaixo é um "operador ternário", que condensa um if/else
    # em uma linha. Leia como: devolva "Direita" SE local for "A",
    # SENÃO devolva "Esquerda".
    return "Direita" if local == "A" else "Esquerda"


# Testando o agente ISOLADAMENTE, sem simulação: percorremos as quatro
# percepções possíveis e conferimos se a decisão faz sentido em cada uma.
# Testar a função sozinha antes de integrá-la é um bom hábito de depuração.
print("Comportamento do agente reativo em cada percepção possível:")
for p in [("A", "Sujo"), ("A", "Limpo"), ("B", "Sujo"), ("B", "Limpo")]:
    print(f"  Percepção {str(p):18s} -> ação: {agente_reativo(p)}")
````

### O Simulador

O cérebro do agente está pronto, agora vamos construir a matriz. O simulador orquestra o laço infinito da simulação de IA:
1. Fornece a **percepção** ao agente.
2. Recebe a **ação**.
3. Atualiza o **estado do mundo**.
4. Atualiza os **pontos**.
5. Guarda o **histórico**.

*(Atenção especial para a cópia do dicionário `mundo` no código. Trabalhar com a referência original causaria vazamento do estado entre diferentes agentes testados).* 

````python
# ---------------------------------------------------------------------
# O SIMULADOR do ambiente.
#
# Ele representa o mundo, entrega percepções ao agente, executa as ações
# escolhidas e acumula a pontuação segundo a medida de desempenho:
#     a cada passo, +1 ponto por local limpo.
# ---------------------------------------------------------------------

def simular(agente, mundo_inicial, local_inicial="A", passos=10, verbose=False):
    """Executa um agente no mundo do aspirador e mede seu desempenho.

    Parâmetros:
        agente (função): o programa do agente a ser testado.
        mundo_inicial (dict): estado inicial, ex. {"A": "Sujo", "B": "Sujo"}.
        local_inicial (str): onde o agente começa.
        passos (int): quantos passos simular.
        verbose (bool): se True, imprime o que acontece em cada passo.

    Devolve:
        tuple: (pontuação total, histórico como lista de dicionários).
    """
    # CÓPIA do dicionário: sem isso, a simulação alteraria o mundo original
    # e o próximo agente testado começaria com o piso já limpo.
    mundo = dict(mundo_inicial)

    local = local_inicial   # posição atual do agente
    memoria = {}            # espaço de estado interno, usado só por alguns agentes
    pontos = 0              # acumulador da pontuação
    historico = []          # registro passo a passo, para inspeção posterior

    # range(1, passos + 1) gera 1, 2, 3, ... até passos.
    # Começamos em 1 apenas para que a numeração exibida fique natural.
    for passo in range(1, passos + 1):

        # PASSO 1 — montar a percepção: onde estou e como está este local.
        percepcao = (local, mundo[local])

        # PASSO 2 — perguntar ao agente. Note que o simulador não sabe
        # NADA sobre como o agente decide: ele só chama a função.
        # É isso que permite trocar de agente sem mexer no simulador.
        acao = agente(percepcao, memoria)

        # PASSO 3 — aplicar a ação, alterando o ambiente ou a posição.
        if acao == "Aspirar":
            mundo[local] = "Limpo"      # a sujeira desaparece deste local
        elif acao == "Direita":
            local = "B"                 # o agente se move
        elif acao == "Esquerda":
            local = "A"
        # "NadaFazer" não cai em nenhum if: nada muda, que é o esperado.

        # PASSO 4 — pontuar. A expressão abaixo percorre os valores do
        # dicionário e soma 1 para cada "Limpo". Lê-se: "some 1 para cada
        # e em mundo.values() tal que e seja igual a Limpo".
        limpos = sum(1 for e in mundo.values() if e == "Limpo")
        pontos += limpos                # += soma ao valor que já estava lá

        # PASSO 5 — registrar. Guardamos um dicionário por passo; depois
        # transformaremos essa lista em DataFrame para visualizar.
        historico.append({
            "passo": passo,
            "percepcao": str(percepcao),
            "acao": acao,
            "limpos": limpos,
        })

        # Impressão opcional, ligada pelo parâmetro verbose.
        if verbose:
            print(f"  passo {passo:2d} | percepção {str(percepcao):18s} | "
                  f"ação {acao:10s} | limpos agora: {limpos}")

    return pontos, historico


print("Simulador definido. Vamos usá-lo na próxima célula.")
````

### Executando a Simulação

Execute o simulador abaixo com os dois ambientes configurados como sujos. 
Observe o log gerado (`verbose=True`) e veja se o robô se comportou conforme suas previsões.

````python
# ---------------------------------------------------------------------
# Primeira simulação: agente reativo, os dois locais sujos, 8 passos.
# verbose=True imprime o que acontece em cada passo.
# ---------------------------------------------------------------------

mundo_sujo = {"A": "Sujo", "B": "Sujo"}

print("Simulação do agente reativo simples:")
pontos, historico = simular(agente_reativo, mundo_sujo, passos=8, verbose=True)

print()
print("Pontuação total:", pontos)

# Máximo teórico: 2 locais limpos × 8 passos = 16 pontos.
# O agente não alcança esse valor porque precisa de alguns passos para
# limpar tudo — e durante esses passos o mundo ainda não estava limpo.
print("Máximo teórico :", 2 * 8)
````

### Transformando o Histórico em Dados (`DataFrame`)

Nosso simulador gerou uma lista de dicionários. No mundo de Ciência de Dados, converter esse log puro em uma tabela estruturada (DataFrame do pandas) nos permite extrair inteligência imediatamente. Faremos isso para analisar a frequência das ações do agente.

### Agente 2 e Agente 3: Explorando Novas Arquiteturas

Vamos definir mais dois perfis de agentes para compará-los cientificamente:

- **Agente Aleatório:** Se vir sujeira, limpa. Se não, toma ações randômicas. Isso quebra laços infinitos em ambientes com restrição de observabilidade.
- **Agente Baseado em Modelo (Com Memória):** Mapeia internamente onde esteve e quais condições observou. Se notar que o mapa inteiro já está limpo, opta pela ação `NadaFazer` ao invés de transitar sem rumo.

> **Dica de Ciência de Dados (A semente):** Usar `random.seed(42)` trava a aleatoriedade. O comportamento randômico será exatamente igual a cada re-execução. Isso garante que o projeto é testável e *reprodutível*!

````python
# ---------------------------------------------------------------------
# AGENTE 2 — aleatório.
# AGENTE 3 — baseado em modelo (com memória).
# ---------------------------------------------------------------------

import random

# Fixa a semente do gerador aleatório: a mesma sequência a cada execução.
# Sem isto, cada rodada daria um resultado diferente e não saberíamos
# se uma mudança nos números veio do código ou do acaso.
random.seed(42)


def agente_aleatorio(percepcao, memoria=None):
    """Aspira quando vê sujeira; caso contrário, move-se ao acaso."""
    local, estado = percepcao

    # A prioridade de aspirar é a mesma do agente reativo.
    if estado == "Sujo":
        return "Aspirar"

    # random.choice() sorteia um elemento da lista com igual probabilidade.
    # É aqui que este agente difere do reativo: ele não segue uma regra
    # fixa de movimento, escolhe ao acaso.
    return random.choice(["Direita", "Esquerda", "NadaFazer"])


def agente_com_memoria(percepcao, memoria):
    """Agente baseado em modelo: mantém estado interno sobre cada local.

    Diferente dos anteriores, este agente ANOTA o que já viu. Quando
    acredita que os dois locais estão limpos, para de se mover.
    """
    local, estado = percepcao

    # ESTADO INTERNO: registramos no dicionário 'memoria' o que sabemos
    # sobre o local atual. Este dicionário sobrevive entre os passos
    # porque é o simulador que o mantém e repassa a cada chamada.
    memoria[local] = estado

    if estado == "Sujo":
        # Vamos aspirar, então já sabemos que o local ficará limpo.
        # Atualizar a memória ANTES da ação é seguro aqui porque a ação
        # é determinística: aspirar sempre limpa.
        memoria[local] = "Limpo"
        return "Aspirar"

    # Descobrimos qual é o OUTRO local, para consultar o que sabemos dele.
    outro = "B" if local == "A" else "A"

    # .get() consulta a memória com segurança: se ainda não visitamos
    # o outro local, devolve None em vez de gerar KeyError.
    if memoria.get(outro) == "Limpo":
        # Os dois locais estão limpos: mover-se seria desperdício.
        return "NadaFazer"

    # Não sabemos como está o outro local: vale a pena ir verificar.
    return "Direita" if local == "A" else "Esquerda"


print("Os três agentes estão definidos:")
print("  1. agente_reativo     - sem memória, regra fixa")
print("  2. agente_aleatorio   - sem memória, movimento ao acaso")
print("  3. agente_com_memoria - com estado interno")
````

### O Grande Experimento Comparativo

Vamos colocar os 3 agentes no simulador, testando as **mesmas regras**, a **mesma quantidade de passos** e os **mesmos cenários** de sujeira. 

Esse é o rigor técnico exigido nos projetos da disciplina.

````python
# ---------------------------------------------------------------------
# EXPERIMENTO COMPARATIVO — o mesmo espírito exigido no Marco 1.
#
# Comparamos os três agentes em três cenários, sob condições idênticas.
# ---------------------------------------------------------------------

# Dicionário que associa um nome legível a cada função de agente.
# Guardar funções dentro de um dicionário é perfeitamente válido em
# Python: funções são valores como quaisquer outros.
agentes = {
    "reativo simples": agente_reativo,
    "aleatório":       agente_aleatorio,
    "com memória":     agente_com_memoria,
}

# Os cenários de teste. Todos os agentes enfrentarão exatamente estes.
cenarios = {
    "ambos sujos":  {"A": "Sujo",  "B": "Sujo"},
    "só A sujo":    {"A": "Sujo",  "B": "Limpo"},
    "ambos limpos": {"A": "Limpo", "B": "Limpo"},
}

random.seed(42)     # semente fixa de novo, para o experimento ser reprodutível

linhas = []         # acumulador dos resultados

# LAÇO EXTERNO: um agente por vez.
for nome_agente, agente in agentes.items():
    # LAÇO INTERNO: para cada agente, todos os cenários.
    for nome_cenario, mundo0 in cenarios.items():
        # O underscore "_" é uma convenção para "não vou usar este valor".
        # Aqui descartamos o histórico e ficamos só com a pontuação.
        pontos, _ = simular(agente, mundo0, passos=10)

        linhas.append({
            "agente": nome_agente,
            "cenário": nome_cenario,
            "pontuação": pontos,
        })

# Lista de dicionários -> DataFrame, o mesmo padrão de antes.
comparativo = pd.DataFrame(linhas)

# .pivot() reorganiza a tabela: os agentes viram linhas, os cenários viram
# colunas e a pontuação preenche as células. Fica bem mais legível que a
# lista corrida de 9 resultados.
tabela_final = comparativo.pivot(index="agente", columns="cenário",
                                 values="pontuação")

print("Pontuação em 10 passos (quanto maior, melhor):")
print(tabela_final)
````
